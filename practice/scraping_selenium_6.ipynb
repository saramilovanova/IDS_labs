```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.common.exceptions import NoSuchElementException, TimeoutException
import time
import pandas as pd
```

```python
MAIN_URL = 'https://med.over.net/forum'

driver = webdriver.Chrome()
driver.get(MAIN_URL)
time.sleep(2)
```

```python
# Main forum page lists different forums. Extract links to all the forums. Also take care of possible ads showing prior to content loading.
'''
[
  {
    "category": "Duševno zdravje in odnosi",
    "forum": "Družina",
    "link": "https://med.over.net/forum/kategorija/dusevno-zdravje-in-odnosi/druzina-3574102/"
    "posts": 85981
  }, 
  {
    "category": "Duševno zdravje in odnosi",
    "forum": "Odnosi in socialna prva pomoč",
    "link": "https://med.over.net/forum/kategorija/dusevno-zdravje-in-odnosi/odnosi-in-socialna-prva-pomoc-3574418/"
    "posts": 58660
  },
  {
    ...
]
'''

def extract_forum_links(driver):
    forum_links = []
    forum_categories = driver.find_elements(By.CLASS_NAME, 'headline-table__text')
    forum_categories = [category.text.strip() for category in forum_categories]
    # print(forum_categories)
    
    tables = driver.find_elements(By.TAG_NAME, 'table')
    print(tables)
    for i, table in enumerate(tables):
        rows = table.find_elements(By.TAG_NAME, 'tr')
        for row in rows:
            forum_name = row.find_element(By.TAG_NAME, 'a')
            name = forum_name.text
            link = forum_name.get_attribute('href')
            posts = row.find_element(By.TAG_NAME, 'span').text
            forum_links.append({
                'category': forum_categories[i],
                'forum': name,
                'link': link,
                'posts': posts
            })
    
    return forum_links
```

```python
# Topics page lists different topics for a specific forum. Extract links to all the topics. Also take care of possible ads showing prior to content loading. Your function should return the following structure:

'''
[
  {
    "topic": "Dodatek za nego otroka, ki potrebuje posebno nego in varstvo nego",
    "answers": 14,
    "views": 564,
    "last-activity": "30.10.2023 ob 12:15",
    "link": "https://med.over.net/forum/tema/dodatek-za-nego-otroka-ki-potrebuje-posebno-nego-in-varstvo-nego-2069589/"
  }, 
  {
    "topic": "Preživnina",
    "answers": 2,
    "views": 19,
    "last-activity": "25.10.2023 ob 22:09",
    "link": "https://med.over.net/forum/tema/prezivnina-5-22508458/"
  },
  {
    ...
]
'''

def extract_topic_links(driver):
    topic_rows = driver.find_elements(By.CLASS_NAME, 'table-forum__row')
    topic_data = []
    for row in topic_rows:
        entries = row.find_elements(By.TAG_NAME, 'td')
        name = entries[0].text
        if name == 'TEMA':
            continue
        link = entries[0].find_element(By.TAG_NAME, 'a').get_attribute('href')
        answers = entries[1].find_element(By.TAG_NAME, 'span').text
        views = entries[2].text
        last_activity = entries[3].text
        topic_data.append({
            'topic': name,
            'answers': answers,
            'views': views,
            'last-activity': last_activity,
            'link': link
        })
        
    return topic_data
```

```python
# Posts parsing
# Posts of a selected forum contain content. Extract posts data. Also take care of possible ads showing prior to content loading.
# Your function should return the following structure:

'''
[
  {
    "user": "petraT",
    "time": "14.10.2007 ob 23:47",
    "content": "Vlogo za dodatek za nego ...", #Extract whole text
  }, 
  {
    "user": "mamica, NEGA",
    "time": "06.12.2011 ob 12:17",
    "content": "Pozdravljeni! Za hčerko prejemam", #Extract whole text
  },
  {
    ...
]
'''

def extract_posts(driver):
    posts = driver.find_elements(By.CLASS_NAME, 'forum-post')
    post_data = []
    for post in posts:
        user_time = post.find_element(By.CLASS_NAME, 'forum-post__name')
        user = user_time.find_elements(By.TAG_NAME, 'span')[0].text
        time = user_time.find_elements(By.TAG_NAME, 'span')[1].text
        # content = post.find_element(By.TAG_NAME, 'p').text
        content = post.find_element(By.CLASS_NAME, 'forum-post__content').text
        post_data.append({
            'user': user,
            'time': time,
            'content': content
        })
        
    return post_data
```

```python
# ## Task 4: End-to-end parsing

# Extract all data without manual supervision. Extract only data from *first pages* (i.e., there is no need to paginate to retrieve all topics, posts). Also take care of possible ads showing prior to content loading.

# The result of parsing should be stored in a pandas data frame, containing the following columns: > category, forum, posts-number, topic-answers, topic-views, topic-last-activity, user, user-post-time, user-content


def extract_data(driver):
    MAIN_URL = 'https://med.over.net/forum'
    driver.get(MAIN_URL)
    time.sleep(2)
    forum_links = extract_forum_links(driver)
    data = []
    for forum_link in forum_links[:2]:
        driver.get(forum_link['link'])
        time.sleep(2)
        topics = extract_topic_links(driver)
        for topic in topics[:2]:
            driver.get(topic['link'])
            time.sleep(2)
            posts = extract_posts(driver)
            for post in posts[:2]:
                data.append({
                    'category': forum_link['category'],
                    'forum': forum_link['forum'],
                    'posts-number': forum_link['posts'],
                    'topic-answers': topic['answers'],
                    'topic-views': topic['views'],
                    'topic-last-activity': topic['last-activity'],
                    'user': post['user'],
                    'user-post-time': post['time'],
                    'user-content': post['content']
                })
    return pd.DataFrame(data)
```

```python
driver = webdriver.Chrome()
data = extract_data(driver)
data
```

```python
# ## Task 5: Data cleaning

# Clean the data in the following way:
# - Remove duplicates
# - Remove rows with missing values
# - Convert columns to appropriate data types
# - Remove leading/trailing whitespaces in columns
# - Remove rows where the user is a bot (i.e., the user contains the word "bot")
# - Remove rows where the user has less than 3 characters


data = pd.read_csv('data/med_over_net.csv')

# Remove duplicates
data = data.drop_duplicates()

# Remove rows with missing values
data = data.dropna()

# Convert columns to appropriate data types
data['topic-answers'] = data['topic-answers'].astype(int)
data['topic-views'] = data['topic-views'].astype(int)

# Remove leading/trailing whitespaces in columns
data['category'] = data['category'].str.strip()
data['forum'] = data['forum'].str.strip()
data['user'] = data['user'].str.strip()
data['user-content'] = data['user-content'].str.strip()

# Remove rows where the user is a bot (i.e., the user contains the word "bot")
data = data[~data['user'].str.contains('bot', case=False)]

# Remove rows where the user has less than 3 characters
data = data[data['user'].str.len() >= 3]

data
```
